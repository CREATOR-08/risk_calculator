```python
import joblib
import os
import re

model_path = r"c:\Users\user\Desktop\deploy model\calibrated_model.pkl"
feature_path = r"c:\Users\user\Desktop\deploy model\feature_names.pkl"

print('model exists:', os.path.exists(model_path))
print('feature file exists:', os.path.exists(feature_path))

model = joblib.load(model_path)
print('model type:', type(model))
print('model feature_names_in_:', getattr(model, 'feature_names_in_', None))

saved_features = joblib.load(feature_path)
print('saved feature_names.pkl:', saved_features)

def normalize_key(key: str) -> str:
    normalized = re.sub(r"[^a-z0-9]+", "_", str(key).strip().lower())
    return normalized.strip("_")

model_features = [str(f) for f in getattr(model, 'feature_names_in_', [])]
print('normalized model features:', [normalize_key(f) for f in model_features])
print('normalized saved features:', [normalize_key(f) for f in saved_features])

print('risk_count in model features?', any(normalize_key(f) == 'risk_count' for f in model_features))
print('risk_count in saved features?', any(normalize_key(f) == 'risk_count' for f in saved_features))

if any(normalize_key(f) == 'risk_count' for f in model_features):
    print('Model still expects risk_count:')
    print(model_features)
else:
    print('Model does not require risk_count based on feature_names_in_.')

if any(normalize_key(f) == 'risk_count' for f in saved_features):
    print('feature_names.pkl still includes risk_count:')
    print(saved_features)
else:
    print('feature_names.pkl does not include risk_count.')
```
